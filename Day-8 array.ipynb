```python
# Count numbers divisible by 3
arr = [34,67,90,24,56,89,23,12,45,78,99,100]
count = 0
for i in arr:
    if i % 3 == 0:
        count += 1
print("Count of numbers divisible by 3:", count)
```

```python
# Print numbers divisible by both 2 and 3
arr = [34,67,90,24,56,89,23,12,45,78,99,100]
for i in arr:
    if i % 2 == 0 and i % 3 == 0:
     print(i)
```

```python
# Find the difference between largest and smallest
arr = [40, 3,57,90,23]
largest = arr[0]
smallest = arr[0]
for i in arr:
    if i > largest:
        largest = i
    if i < smallest:
        smallest = i
difference = largest - smallest
print("Difference between largest and smallest:", difference)
```
