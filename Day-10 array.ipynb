```python
# Find the last even number
arr = [1, 3, 5, 7, 8, 10, 11]
last_even = None
for i in arr:
    if i% 2 == 0:
        last_even = i
print(last_even)
```

```python
# Find the last odd number
arr = [1, 3, 5, 7, 8, 10, 11]
last_odd = None
for i in arr:
    if i% 2 != 0:
        last_odd = i
print(last_odd)
```
